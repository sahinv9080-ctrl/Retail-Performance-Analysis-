```python
# data type
# Convert to datetime

df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    format="mixed",
    dayfirst=True
)

df['Ship Date'] = pd.to_datetime(
    df['Ship Date'],
    format="mixed",
    dayfirst=True
)

df.head()
```

```python
# Convert to MySQL format (YYYY-MM-DD)

df["Order Date"]=df["Order Date"].dt.strftime("%Y-%m-%d")
df["Ship Date"]=df["Ship Date"].dt.strftime("%Y-%m-%d")
print(df.dtypes)

print(df["Order Date"].sample(20).tolist())
print(df["Ship Date"].sample(20).tolist())






```

```python
df.columns = (
    df.columns
      .str.strip()        # removes leading/trailing spaces
      .str.lower()        # converts all column names to lowercase
      .str.replace(" ", "_")  # replaces spaces with underscores
      .str.replace("-", "_")  # replaces hyphens with underscores
)
```

```python

  #x is horizontal boxplot of sales
#It lets you arrange multiple plots in one figure(subplot)
#detect the outlier (boxplot)
#loop + range(enumerate)

numerical_columns = ["sales", "profit", "quantity", "discount"]

plt.figure(figsize=(15,10))
for i, plot in enumerate(numerical_columns, 1):
    plt.subplot(3,3,i)
    sns.boxplot(x=df[plot], color="lightblue")
    plt.title(f"Boxplot of {plot}")
    plt.xlabel(plot)

plt.tight_layout()
plt.show()

```

```python
import matplotlib.pyplot as plt

plt.bar(df["region"], df["sales"], color="g")
plt.grid()
plt.title("Sales trend")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.show()
```

```python
# Sum profit for each region
top10 = (
    df.groupby("region")["profit"].sum().sort_values(ascending=False).head(10))

plt.figure(figsize=(8,8))
plt.pie(
    top10.values,
    labels=top10.index,
    autopct='%1.2f%%',
    startangle=90
)

plt.title("Top 10 Region by Profit")
plt.axis('equal')
plt.show()
```

```python
plt.scatter(df["categories"],df["sales"],color="g")
plt.grid()
plt.title("Sales trend")
plt.xlabel("region")
plt.ylabel("Sales")
plt.show()
```

```python
sales_trend = df.groupby("order_date")["sales"].sum()

plt.figure(figsize=(10,5))
plt.plot(sales_trend.index, sales_trend.values)
plt.title("Sales Trend Over Time")
plt.xlabel("Order Date")
plt.ylabel("Sales")

plt.show()
```
